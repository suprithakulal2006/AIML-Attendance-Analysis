#### Section 1: Data Inspection

```python
"""Q1. Display the first 10 rows, the shape of the DataFrame, and the data types of each column."""
import pandas as pd

df = pd.read_excel("database/aiml_attendance.xlsx")

print("First 10 rows:")
display(df.head(10))

print("Shape of the DataFrame:")
print(df.shape)

print("Data types of each column:")
print(df.dtypes)
```

```python
"""Q2. Check for missing values in each column. Identify which columns have null values (especially Last Present Date with "Never") and report the count. """
print("Missing values in each column:")
print(df.isnull().sum())

print("\nColumns with null values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nCount of 'Never' in Last Present Date:")
print(df["Last Present Date"].astype(str).str.strip().eq("Never").sum())
```

```python
"""Q3. Rename the columns to snake_case (e.g., Roll No → roll_no, Student Name → student_name, Attendance % → attendance_pct). Display the updated column list."""
import re

df.columns = [
    re.sub(r'[^a-zA-Z0-9]+', '_', col.strip()).strip('_').lower()
    for col in df.columns
]

print("Updated column list:")
print(df.columns.tolist())
```

```python
"""Q4. Convert Attendance % from string format (e.g., "64%") to a numeric float column (0.64 or 64.0). Also convert Last Present Date to datetime, handling "Never" as NaT."""

print(df.columns.tolist())

attendance_col = next(col for col in df.columns if 'attendance' in col.lower())
date_col = next(col for col in df.columns if 'last present date' in col.lower().replace('_', ' '))

df[attendance_col] = pd.to_numeric(
    df[attendance_col].astype(str).str.replace('%', '', regex=False),
    errors='coerce'
)

df[date_col] = pd.to_datetime(
    df[date_col].replace('Never', pd.NaT),
    errors='coerce'
)

print(df.dtypes)
display(df.head())
```

#### Section 2: Filtering & Selection

```python
"""Q5. Filter and display all students whose Status is "Critical". Sort them by Attendance % in ascending order."""

print(df.columns.tolist())

status_col = next(col for col in df.columns if col.strip().lower() == "status")
attendance_col = next(col for col in df.columns if "attendance" in col.strip().lower())

critical_students = df[
    df[status_col].astype(str).str.strip().str.lower() == "critical"
].sort_values(by=attendance_col, ascending=True)

display(critical_students)
```

```python
"""Q6. Extract all students who have 0 present days (fully absent). List their roll numbers, names, and consecutive absences."""

print(df.columns.tolist())

fully_absent = df[pd.to_numeric(df["present"], errors="coerce") == 0]

print("Students with 0 present days:")
display(fully_absent[["roll_no", "student_name", "consecutive_absences"]])
```

```python
"""Q7. Using .loc and .iloc, display:
(a) All "Safe" students using .loc
(b) The first 15 rows and columns 2-6 using .iloc """

print("All Safe students using .loc:")
display(df.loc[df["status"].str.strip().str.lower() == "safe"])

print("First 15 rows and columns 2-6 using .iloc:")
display(df.iloc[:15, 1:6])
```

```python
"""Q8. Find all students whose Consecutive Absences is greater than or equal to 5. Display their name, status, and last present date."""

students = df[pd.to_numeric(df["consecutive_absences"], errors="coerce") >= 5]

print("Students with 5 or more consecutive absences:")
display(students[["student_name", "status", "last_present_date"]])
```

### Section 3: Grouping & Aggregation 

```python
"""Q9. Group the data by Status and compute: 
a) Count of students 
b) Average attendance %  
c) Maximum consecutive absences"""

status_summary = df.groupby("status").agg(
    student_count=("student_name", "count"),
    average_attendance=("attendance", "mean"),
    max_consecutive_absences=("consecutive_absences", "max")
)

print("Summary by Status:")
display(status_summary)
```

```python
"""Q10. Create a new column Attendance Category:
≥ 90% → "Excellent"
75-89% → "Good"
60-74% → "Average"
< 60% → "Poor"
Then use groupby() to count students in each category."""

df["attendance"] = pd.to_numeric(
    df["attendance"].astype(str).str.replace("%", "", regex=False),
    errors="coerce"
)

df["Attendance Category"] = pd.cut(
    df["attendance"],
    bins=[-float("inf"), 60, 75, 90, float("inf")],
    labels=["Poor", "Average", "Good", "Excellent"],
    right=False
)

df["Attendance Category"] = df["Attendance Category"].replace({
    "Good": "Good",
    "Average": "Average"
})

print("Student count in each Attendance Category:")
display(df.groupby("Attendance Category", observed=False)["student_name"].count())
```

```python
"""Q11. Calculate the total number of classes conducted (sum of Total Classes) and the total present/absent counts across all students using .agg()."""

totals = df.agg({
    "total_classes": "sum",
    "present": "sum",
    "absent": "sum"
})

print("Total classes conducted, present count, and absent count:")
display(totals)
```

```python
"""Q12. Group by Status and use .agg() to find:
a)min, max, and mean of Present
b)sum of Absent"""

status_summary = df.groupby("status").agg(
    present_min=("present", "min"),
    present_max=("present", "max"),
    present_mean=("present", "mean"),
    absent_sum=("absent", "sum")
)

display(status_summary)
```

### Section 4: Feature Engineering & Transformation

```python
"""Q13. Create a new column Risk Score using the formula:
Risk Score = (Absent * 0.4) + (Consecutive Absences * 0.6)
Display the top 10 students with the highest risk score."""

df["risk_score"] = (
    df["absent"] * 0.4
    + df["consecutive_absences"] * 0.6
)

top_10_risk = df.sort_values(
    by="risk_score",
    ascending=False
).head(10)

print("Top 10 students with the highest risk score:")
display(top_10_risk[["roll_no", "student_name", "absent", "consecutive_absences", "risk_score"]])

```

```python
"""Q14. Add a column Classes Needed (Numeric) by extracting the numeric value from the Classes Needed (75%) column (e.g., "5 more" → 5, "✓ Met" → 0)."""

df["classes_needed_numeric"] = (
    df["classes_needed_75"]
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
    .fillna("0")
    .astype(int)
)

display(df[["student_name", "classes_needed_75", "classes_needed_numeric"]])
```

```python
"""Q15. Using apply() with a lambda function, create a column Attendance Flag that returns:
a)"Defaulter" if attendance < 75%
b)"Safe" otherwise """

df["Attendance Flag"] = df["attendance"].apply(
    lambda x: "Defaulter" if x < 75 else "Safe"
)

display(df[["student_name", "attendance", "Attendance Flag"]])
```

```python
"""Q16. Use pd.cut() to bin students into attendance ranges: [0-50, 50-75, 75-90, 90-100] and label them as ["Critical", "Warning", "Good", "Excellent"]."""

df["Attendance Range"] = pd.cut(
    df["attendance"],
    bins=[0, 50, 75, 90, 100],
    labels=["Critical", "Warning", "Good", "Excellent"],
    include_lowest=True
)

display(df[["student_name", "attendance", "Attendance Range"]])
```

### Section 5: Advanced Analysis 

```python
"""Q17. Using pivot_table(), create a summary showing the count of students by Status and Attendance Category (created in Q10)."""

pivot_summary = pd.pivot_table(
    df,
    index="status",
    columns="Attendance Category",
    values="student_name",
    aggfunc="count",
    fill_value=0,
    observed=False
)

print("Count of students by Status and Attendance Category:")
display(pivot_summary)
```

```python
"""Q18. Identify duplicate student names (e.g., "ANANYA", "DISHA") and display all rows where the base name repeats. Use string manipulation to extract the base name."""

df["base_name"] = df["student_name"].str.replace(r"\s*\d+$", "", regex=True).str.strip().str.upper()

duplicate_names = df[df.duplicated(subset="base_name", keep=False)].sort_values("base_name")

display(duplicate_names)
```

```python
""""Q19. Merge this dataset with a hypothetical Student_Details DataFrame containing Roll No, Branch, and Semester. Perform an inner join and display the merged result."""

student_details = pd.DataFrame({
    "roll_no": df["roll_no"].unique(),
    "branch": ["AIML"] * df["roll_no"].nunique(),
    "semester": [5] * df["roll_no"].nunique()
})

merged_df = pd.merge(
    df,
    student_details,
    on="roll_no",
    how="inner"
)

print("Merged Student Data:")
display(merged_df)
```

```python
"""Q20. Export the following to separate CSV files:
(a) All "Critical" students
(b) Summary statistics grouped by Status
(c) Top 10 students by Risk Score"""

critical_students = df[df["status"].str.strip().str.lower() == "critical"]
critical_students.to_csv("critical_students.csv", index=False)

status_summary = df.groupby("status").agg(
    student_count=("student_name", "count"),
    average_attendance=("attendance", "mean"),
    max_consecutive_absences=("consecutive_absences", "max")
)
status_summary.to_csv("status_summary.csv")

top_10_risk = df.sort_values(by="risk_score", ascending=False).head(10)
top_10_risk.to_csv("top_10_risk_score.csv", index=False)

print("CSV files exported successfully:")
print("1. critical_students.csv")
print("2. status_summary.csv")
print("3. top_10_risk_score.csv")
```
